```python
#Suffix-Adding Function

def add_suffix(rows, column):
    for i in rows:
        if i[column] is not None and i[column] != "":
            i[column] += "_88"
    return rows
```

# Create & fill SQL tables

```python
# Create & fill SQL table for Item
create_item_table_query = """
DROP TABLE IF EXISTS item;
CREATE TABLE item (
    Item_ID VARCHAR(255) PRIMARY KEY,
    Item_Description VARCHAR(255),
    Item_Price INTEGER,
    Item_Weight INTEGER,
    Item_AnnualSales INTEGER,
    Item_StockLocation VARCHAR(255)
);
"""
for statement in create_item_table_query.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

insert_item_query = """
INSERT INTO item (
    Item_ID, Item_Description, Item_Price, Item_Weight,
    Item_AnnualSales, Item_StockLocation
) VALUES (%s, %s, %s, %s, %s, %s)
"""

cursor.executemany(insert_item_query, [
    (
        r["Item_ID"], r["Item_Description"], r["Item_Price"],
        r["Item_Weight"], r["Item_AnnualSales"], r["Item_StockLocation"]
    )
    for r in item_rows
])

conn.commit()
print("item table created & filled.")
```

```python
# Create & fill Cases SQL table
create_cases_table_query = """
DROP TABLE IF EXISTS cases;
CREATE TABLE cases (
    Case_ID CHAR(3) PRIMARY KEY,
    Item_Quantity INTEGER,
    Case_Weight INTEGER
);
"""
for statement in create_cases_table_query.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)
        
insert_cases_query = """
INSERT INTO cases (Case_ID, Item_Quantity, Case_Weight)
VALUES (%s, %s, %s)
"""

cursor.executemany(insert_cases_query, [
    (r["Case_ID"], r["Item_Quantity"], r["Case_Weight"])
    for r in cases_rows
])

conn.commit()
print("cases table created & filled.")
```

```python
# Create & fill Customer SQL table
#Customer_Num = varchar(5+3) because of added "_88"
create_customer_table_query = """
DROP TABLE IF EXISTS customer;
CREATE TABLE customer (
    Customer_Num VARCHAR(8) PRIMARY KEY,
    Customer_FirstName VARCHAR(255),
    Customer_LastName VARCHAR(255),
    Customer_Address VARCHAR(255),
    Customer_City VARCHAR(255),
    Customer_State CHAR(2),
    Customer_ZIP CHAR(5),
    Customer_Phone CHAR(10),
    Customer_Email VARCHAR(255)
);
"""
for statement in create_customer_table_query.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

insert_customer_query = """
INSERT INTO customer (
    Customer_Num, Customer_FirstName, Customer_LastName,
    Customer_Address, Customer_City, Customer_State,
    Customer_ZIP, Customer_Phone, Customer_Email
) VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

cursor.executemany(insert_customer_query, [
    (
        r["Customer_Num"], r["Customer_FirstName"],
        r["Customer_LastName"], r["Customer_Address"],
        r["Customer_City"], r["Customer_State"],
        r["Customer_ZIP"], r["Customer_Phone"], r["Customer_Email"]
    )
    for r in cust_rows
])

conn.commit()
print("customer table created & filled.")
```

```python
# Create & fill CustomerOrder SQL table
create_customer_order_table_query = """
DROP TABLE IF EXISTS customerOrder;
CREATE TABLE customerOrder (
    Order_Num CHAR(8) PRIMARY KEY,
    Customer_Num VARCHAR(8),
    Order_Date DATE,
    Order_ReqShipDate DATE,
    Order_ActualShipDate DATE,
    Order_Carrier VARCHAR(255),
    
    FOREIGN KEY (Customer_Num) 
        REFERENCES customer(Customer_Num)
        ON UPDATE CASCADE
        ON DELETE CASCADE
);
"""
for statement in create_customer_order_table_query.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

insert_customer_order_query = """
INSERT INTO customerOrder (
    Order_Num, Customer_Num, Order_Date,
    Order_ReqShipDate, Order_ActualShipDate, Order_Carrier
) VALUES (%s, %s, %s, %s, %s, %s)
"""

cursor.executemany(insert_customer_order_query, [
    (
        r["Order_Num"], r["Customer_Num"], r["Order_Date"],
        r["Order_ReqShipDate"], r["Order_ActualShipDate"],
        r["Order_Carrier"]
    )
    for r in custOrd_rows
])

conn.commit()
print("customerOrder table created & filled.")
```

```python
# Create & fill Order Item SQL table
create_order_item_table_query = """
DROP TABLE IF EXISTS orderItem;
CREATE TABLE orderItem (
    Order_Num CHAR(8),
    Item_ID VARCHAR(255),
    Item_Qty INTEGER,
    PRIMARY KEY (Order_Num, Item_ID),
    
    FOREIGN KEY (Order_Num) 
        REFERENCES customerOrder(Order_Num)
        ON UPDATE CASCADE
        ON DELETE CASCADE,
        
    FOREIGN KEY (Item_ID) 
        REFERENCES item(Item_ID)
);
"""
for statement in create_order_item_table_query.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

insert_order_item_query = """
INSERT INTO orderItem (Order_Num, Item_ID, Item_Qty)
VALUES (%s, %s, %s)
"""

cursor.executemany(insert_order_item_query, [
    (r["Order_Num"], r["Item_ID"], r["Item_Qty"])
    for r in ordItem_rows
])
conn.commit()
print("order_item table created & filled.")
```

```python
# Create & fill Vendor SQL table
create_vendor_table_query = """
DROP TABLE IF EXISTS vendor;
CREATE TABLE vendor (
    Vendor_Num CHAR(5) PRIMARY KEY,
    Vendor_Name VARCHAR(255),
    Vendor_Address VARCHAR(255),
    Vendor_Phone CHAR(10),
    Vendor_Email VARCHAR(255),
    Vendor_City VARCHAR(255),
    Vendor_State CHAR(2),
    Vendor_ZIP CHAR(5)
);
"""
for statement in create_vendor_table_query.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

insert_vendor_query = """
INSERT INTO vendor (
    Vendor_Num, Vendor_Name, Vendor_Address, Vendor_Phone,
    Vendor_Email, Vendor_City, Vendor_State, Vendor_ZIP
) VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
"""

cursor.executemany(insert_vendor_query, [
    (
        r["Vendor_Num"], r["Vendor_Name"], r["Vendor_Address"],
        r["Vendor_Phone"], r["Vendor_Email"], r["Vendor_City"],
        r["Vendor_State"], r["Vendor_ZIP"]
    )
    for r in vendor_rows
])

conn.commit()
print("vendor table created & uploaded.")
```

```python
# Create & fill Shipment SQL table
create_shipment_table_query = """
DROP TABLE IF EXISTS shipment;
CREATE TABLE shipment (
    Shipment_Num CHAR(9) PRIMARY KEY,
    Vendor_Num CHAR(5),
    Shipment_ReceiptDate DATE,
    Shipment_NumPallets INTEGER,
    PurchaseOrder_Num VARCHAR(255),
    
    FOREIGN KEY (Vendor_Num) 
        REFERENCES vendor(Vendor_Num)
        ON UPDATE CASCADE
        ON DELETE CASCADE
);
"""
for statement in create_shipment_table_query.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)
        
insert_shipment_query = """
INSERT INTO shipment (
    Shipment_Num, Vendor_Num, Shipment_ReceiptDate,
    Shipment_NumPallets, PurchaseOrder_Num
) VALUES (%s, %s, %s, %s, %s)
"""

cursor.executemany(insert_shipment_query, [
    (
        r["Shipment_Num"], r["Vendor_Num"],
        r["Shipment_ReceiptDate"], r["Shipment_NumPallets"],
        r["PurchaseOrder_Num"]
    )
    for r in shipment_rows
])

conn.commit()
print("shipment table created & filled.")
```

```python
# Create & fill Shipment Info SQL table
create_shipment_info_table_query = """
DROP TABLE IF EXISTS shipmentInfo;
CREATE TABLE shipmentInfo (
    Shipment_Num CHAR(9),
    Case_ID CHAR(3),
    Item_ID VARCHAR(255),
    Pallet_ID VARCHAR(3),
    Item_Qty INTEGER,
    PRIMARY KEY (Shipment_Num, Case_ID),
    
    FOREIGN KEY (Shipment_Num) 
        REFERENCES shipment(Shipment_Num)
        ON UPDATE CASCADE
        ON DELETE CASCADE,
        
    FOREIGN KEY (Case_ID) 
        REFERENCES cases(Case_ID)
        ON UPDATE CASCADE
        ON DELETE CASCADE,
        
    FOREIGN KEY (Item_ID) 
        REFERENCES item(Item_ID)
        ON UPDATE CASCADE
        ON DELETE CASCADE,
        
    FOREIGN KEY (Pallet_ID) 
        REFERENCES pallet(Pallet_ID)
        ON UPDATE CASCADE
        ON DELETE CASCADE
);
"""
for statement in create_shipment_info_table_query.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

insert_shipment_info_query = """
INSERT INTO shipmentInfo (
    Shipment_Num, Case_ID, Item_ID, Pallet_ID, Item_Qty
) VALUES (%s, %s, %s, %s, %s)
"""

cursor.executemany(insert_shipment_info_query, [
    (
        r["Shipment_Num"], r["Case_ID"], r["Item_ID"],
        r["Pallet_ID"], r["Item_Qty"]
    )
    for r in shipInfo_rows
])

conn.commit()
print("shipmentInfo table created & filled.")
```

```python
query1 = """
SELECT item.Item_ID, item.Item_Description, orderItem.Item_Qty
FROM item
JOIN orderItem ON item.Item_ID = orderItem.Item_ID
JOIN customerOrder ON orderItem.Order_Num = customerOrder.Order_Num
WHERE customerOrder.Customer_Num = "C25_88"; """

cursor.execute(query1)
for row in cursor.fetchall():
    print(row)    
```

# Part 4: My Questions

```python
# Question 1
# When was Bob Felton's first order with us placed and when was his last?

queryQ1 = """
SELECT MIN(co.Order_Date), MAX(co.Order_Date)
FROM customerOrder co
JOIN customer c USING (Customer_Num)
WHERE c.Customer_FirstName = 'Bob' AND c.Customer_LastName = 'Felton'; """

cursor.execute(queryQ1)
for row in cursor.fetchall():
    print(row)
```

```python
# Question 2
# When did the last shipment with Footwear arrive and when does the next shipment with Footwear need to leave?

queryQ2 = """
SELECT 
    (SELECT MAX(s.Shipment_ReceiptDate)
        FROM shipment s
        JOIN shipmentInfo USING (Shipment_Num)
        JOIN item i USING (Item_ID)
        WHERE i.Item_Description = 'Footwear') AS lastArrived,
    (SELECT MAX(co.Order_ReqShipDate)
        FROM customerOrder co
        JOIN orderItem USING (Order_Num)
        JOIN item i USING (Item_ID)
        WHERE i.Item_Description = 'Footwear') AS nextShipment;
"""


cursor.execute(queryQ2)
for row in cursor.fetchall():
    print(row)
```

```python
# Question 3
# What is the total dollar value each customer has spent on our orders, in descending order?

queryQ3 = """
SELECT c.Customer_FirstName, c.Customer_LastName, 
    SUM(i.Item_Price * oi.Item_Qty) AS totalDollars
FROM customer c
JOIN customerOrder co USING (Customer_Num)
JOIN orderItem oi USING (Order_Num)
JOIN item i USING (Item_ID)
GROUP BY c.Customer_FirstName, c.Customer_LastName
ORDER BY totalDollars DESC;"""

cursor.execute(queryQ3)
for row in cursor.fetchall():
    print(row)
```

```python
# Question 4
# Who has delivered shipments to us with overweight (weight >= 20) cases?

queryQ4 = """
SELECT DISTINCT v.Vendor_Name
FROM vendor v
JOIN shipment USING (Vendor_Num)
JOIN shipmentInfo USING (Shipment_Num)
JOIN cases c USING (Case_ID)
WHERE c.Case_Weight >= 20; """

cursor.execute(queryQ4)
for row in cursor.fetchall():
    print(row)
```
