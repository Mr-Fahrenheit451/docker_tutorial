# Loading Customer.csv into MySQL

This notebook reads `data/orders/Customer.csv` and writes it into the `iem4723_mysql_tutorial` database running in the `mysql` container defined in `docker-compose.yml`. This version uses plain `pymysql` cursors and Python's built-in `csv` module, no pandas at all. If you want to see the same thing done with pandas DataFrames instead, check out `mysql_test_pandas.ipynb` sitting right next to this one.

Run this notebook from inside the `jupyter` service (it already has `pymysql` installed, and connects to the `mysql` service by its container name). If you are instead running this notebook directly on your host machine, change `DB_HOST` below to `localhost`, since port 3306 is also published to the host.

```python
import pymysql
import csv
```

**REMEMBER:** You need to connect to the database instance thats running on one of your container images. Therefore, you need to provide details to Python on how to connect to it. This is similar to netowrking and ports. In this case we provide 3306 as our port since thats the port our MySQL instance is *listening* on. 

```python
DB_HOST = "mysql"          # use "localhost" if running outside the jupyter container
DB_PORT = 3306
DB_USER = "root"
DB_PASSWORD = "OSUGoPokes"
DB_NAME = "iem4723_mysql_tutorial"
```

Now we use these details to actually open a connection to the database. *conn* represents our live connection to the DB instance, and *cursor* is what lets us actually send commands through that connection and read back whatever comes back. We will use both of these again and again below.

```python
conn = pymysql.connect(
    host=DB_HOST,
    port=DB_PORT,
    user=DB_USER,
    password=DB_PASSWORD,
    database=DB_NAME,
)
cursor = conn.cursor()
```

We have ensured that the `data` folder is already volume mounted. As a result, all the files in data should technically show up here.

```python
csv_path = "../data/orders/Customer.csv"

with open(csv_path, newline="") as f:
    reader = csv.DictReader(f)
    customer_rows = list(reader)

customer_rows[:3]
```

Notice that every value in *customer_rows* is just a plain string, even things like `Customer_ZIP` and `Customer_Phone`. The `csv` module never tries to guess whether something is a number, so we never have to worry about losing a leading zero off a zip code the way we would with a spreadsheet tool.

Now we need to create a table that matches these columns before we can put any of this data into MySQL. Notice how we use VARCHAR for pretty much everything, since we dont really need to do any *arithmetic* on any of these fields, only look them up.

For that we will first create a query string `create_customer_table`

```python
#this is our query string
create_customer_table = """
DROP TABLE IF EXISTS customer;
CREATE TABLE customer (
    Customer_Num VARCHAR(10) PRIMARY KEY,
    Customer_FirstName VARCHAR(50),
    Customer_LastName VARCHAR(50),
    Customer_Address VARCHAR(100),
    Customer_City VARCHAR(50),
    Customer_State VARCHAR(2),
    Customer_ZIP VARCHAR(10),
    Customer_Phone VARCHAR(15),
    Customer_Email VARCHAR(100)
);
"""

for statement in create_customer_table.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

conn.commit()
print("customer table created.")
```

Now we have every row from Customer.csv sitting in *customer_rows*, and an empty *customer* table ready to receive them. How do we get the data in? We loop through *customer_rows* one row at a time, and run an INSERT for each one. Notice the `%s` placeholders in the query below. *cursor.execute* fills those in for us safely, so we never have to paste raw text directly into a SQL string ourselves.

```python
insert_customer = """
INSERT INTO customer (Customer_Num, Customer_FirstName, Customer_LastName, Customer_Address, Customer_City, Customer_State, Customer_ZIP, Customer_Phone, Customer_Email)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s);
"""

for row in customer_rows:
    cursor.execute(insert_customer, (
        row["Customer_Num"],
        row["Customer_FirstName"],
        row["Customer_LastName"],
        row["Customer_Address"],
        row["Customer_City"],
        row["Customer_State"],
        row["Customer_ZIP"],
        row["Customer_Phone"],
        row["Customer_Email"],
    ))

conn.commit()
print(f"Wrote {len(customer_rows)} rows to the customer table.")
```

Now try doing a SELECT from the table you just wrote. It should present everything back to you.

```python
# Verify by reading the data back from MySQL
cursor.execute("SELECT * FROM customer;")
for row in cursor.fetchall():
    print(row)
```

## Replicating the phpMyAdmin Steps in Python

Remember everything we just did for customer through phpMyAdmin? Creating a table, inserting a row by hand, importing a CSV, filtering with WHERE and AND? Turns out we can do all of that straight from Python too, using the exact same *conn* and *cursor* we already opened above. We will walk through the same steps again below, except this time using Vendor.csv instead of Customer.csv, just so you can see the same tricks work on a completely different file.

### Creating a Table

How did we create a table in phpMyAdmin again? We typed in column names and types by hand on the Structure tab. So how do we do the same thing in Python? We just write out that same CREATE TABLE statement as a string, and run it through *cursor*, same as always.

```python
create_vendor_manual = """
DROP TABLE IF EXISTS vendor_manual;
CREATE TABLE vendor_manual (
    Vendor_Num VARCHAR(10) PRIMARY KEY,
    Vendor_Name VARCHAR(50),
    Vendor_City VARCHAR(50),
    Vendor_State VARCHAR(2)
);
"""

for statement in create_vendor_manual.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

conn.commit()
print("vendor_manual table created.")
```

### Inserting Data via SQL

Remember the INSERT INTO ... VALUES statement we ran directly in phpMyAdmin's SQL tab? Same idea here. We write out the exact same statement as a plain string and run it through *cursor*, the same way we just ran CREATE TABLE above.

```python
insert_vendor = """
INSERT INTO vendor_manual (Vendor_Num, Vendor_Name, Vendor_City, Vendor_State)
VALUES ('V1107', 'CALATLANTIC GROUP', 'ARLINGTON', 'VA');
"""

cursor.execute(insert_vendor)
conn.commit()

cursor.execute("SELECT * FROM vendor_manual;")
for row in cursor.fetchall():
    print(row)
```

### Importing a CSV File

How did phpMyAdmin's Import tab build a table for us without us writing any SQL? It just read the CSV's header row and figured out the columns on its own. We dont get that shortcut in plain Python, so we do it in two steps instead. First we create the vendor table ourselves, matching the columns in Vendor.csv. Then we loop through the file and insert one row at a time, the exact same way we just inserted a single row above.

```python
create_vendor_table = """
DROP TABLE IF EXISTS vendor;
CREATE TABLE vendor (
    Vendor_Num VARCHAR(10) PRIMARY KEY,
    Vendor_Name VARCHAR(50),
    Vendor_Address VARCHAR(100),
    Vendor_Phone VARCHAR(15),
    Vendor_Email VARCHAR(100),
    Vendor_City VARCHAR(50),
    Vendor_State VARCHAR(2),
    Vendor_ZIP VARCHAR(10)
);
"""

for statement in create_vendor_table.strip().split(";"):
    if statement.strip():
        cursor.execute(statement)

#this is a simple csv file reading in Python.
with open("../data/orders/Vendor.csv", newline="") as f:
    reader = csv.DictReader(f)
    vendor_rows = list(reader)

insert_vendor_row = """
INSERT INTO vendor (Vendor_Num, Vendor_Name, Vendor_Address, Vendor_Phone, Vendor_Email, Vendor_City, Vendor_State, Vendor_ZIP)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s);
"""

for row in vendor_rows:
    cursor.execute(insert_vendor_row, (
        row["Vendor_Num"],
        row["Vendor_Name"],
        row["Vendor_Address"],
        row["Vendor_Phone"],
        row["Vendor_Email"],
        row["Vendor_City"],
        row["Vendor_State"],
        row["Vendor_ZIP"],
    ))

conn.commit()
print(f"Imported {len(vendor_rows)} rows into the vendor table.")
```

### Querying a Table

Remember running SELECT * FROM ... WHERE 1 in phpMyAdmin's SQL tab? That just returned every row in the table. Whats the Python equivalent? We run the exact same query through *cursor*, then call *cursor.fetchall()* to grab every row it found.

```python
cursor.execute("SELECT * FROM vendor;")
for row in cursor.fetchall():
    print(row)
```

### Filtering with WHERE

Remember adding Item_Price >= 3 in the SQL tab earlier to filter our results? Same idea here. We filter vendor down to only the vendors based in Texas.

```python
cursor.execute("SELECT * FROM vendor WHERE Vendor_State = 'TX';")
for row in cursor.fetchall():
    print(row)
```

### Combining Conditions with AND

Remember filtering for items priced at three dollars or more that were also Slacks? Same trick again. We narrow our Texas vendors down even further to only the ones based in Dallas.

```python
cursor.execute("SELECT * FROM vendor WHERE Vendor_State = 'TX' AND Vendor_City = 'DALLAS';")
for row in cursor.fetchall():
    print(row)
```

### Closing Up

Same idea as closing a file when you're done with it. We close *cursor* first, then *conn*, so the database knows we are finished and can free up the connection.

```python
cursor.close()
conn.close()
```
